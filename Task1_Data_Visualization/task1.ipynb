## Scatter Plot
Relationship between Petal Length and Petal Width

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=iris,
    x="petal_length",
    y="petal_width",
    hue="species"
)

plt.title("Petal Length vs Petal Width")

plt.savefig("images/scatter_plot.png")

plt.show()
```

### Observation, Histogram

The scatter plot shows a strong relationship between petal length and petal width. Different species form distinct clusters.

```python
iris.hist(figsize=(10,8))

plt.suptitle("Distribution of Iris Features")

plt.savefig("images/histogram.png")

plt.show()
```

## Box Plot
Detecting Outliers

```python
plt.figure(figsize=(10,6))

sns.boxplot(data=iris)

plt.title("Box Plot of Iris Features")

plt.savefig("images/box_plot.png")

plt.show()
```
